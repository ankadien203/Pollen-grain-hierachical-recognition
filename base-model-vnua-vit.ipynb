```python
# CELL 1: import needed libraries
import os, random, glob
from pathlib import Path
import numpy as np
import pandas as pd
from PIL import Image
import timm
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from torchvision import models, transforms
import torch.nn.functional as F

import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay


IMG_ROOT   = "/kaggle/input/datasets/ichbingiangthanh/vnua-dataset-fullnamespieces-080626/crop_data - filter_250126 - Full name/crop_data"     
TAX_XLSX   = "/kaggle/input/datasets/ichbingiangthanh/id-vnua-080626/id_vnua.xlsx"  
TAX_SHEET  = "Sheet3"

SEED = 42
#VAL_RATIO = 0.2
IMG_SIZE = 224
BATCH_SIZE = 32
EPOCHS = 10
LR = 3e-4
WD = 1e-4

# trọng số loss theo tầng
W_O, W_F, W_G, W_S = 1.0, 0.8, 0.6, 0.4

SAVE_PATH = "/kaggle/working/trained_model_vnua_base_ViT.pt"
CM_SAVE_PATH = "/kaggle/working/species_confusion_matrix_ViT.png"

IMG_EXT = {".jpg",".jpeg",".png",".bmp",".tif",".tiff",".webp"}

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
print("Device:", DEVICE)
```

```python
# cell 2: read excel file which has 4 column sp, genus, family, order and rename to eng
#labling each image with sp as a key in a dict named by_spieces
def norm_str(x):
    return " ".join(str(x).strip().split())

tax = pd.read_excel(TAX_XLSX, sheet_name=TAX_SHEET)
tax.columns = [norm_str(c).lower() for c in tax.columns]

rename_map = {}
if "loài" in tax.columns: rename_map["loài"] = "species"
if "chi"  in tax.columns: rename_map["chi"]  = "genus"
if "họ"   in tax.columns: rename_map["họ"]   = "family"
if "bộ"   in tax.columns: rename_map["bộ"]   = "order"
tax = tax.rename(columns=rename_map)

need = ["species","genus","family","order"]
missing = [c for c in need if c not in tax.columns]
assert not missing, f"Taxonomy thiếu cột {missing}. Columns hiện có: {list(tax.columns)}"

tax = tax[need].copy()
for c in need:
    tax[c] = tax[c].apply(norm_str)

tax_species_set = set(tax["species"].tolist())

species_to_genus = dict(zip(tax["species"], tax["genus"]))
genus_to_family  = dict(zip(tax["genus"],  tax["family"]))
family_to_order  = dict(zip(tax["family"], tax["order"]))

print("Tax rows:", len(tax))
# scan images & keep only species existing in taxonomy
samples = []
root = Path(IMG_ROOT)
print(root.iterdir())
assert root.exists(), f"IMG_ROOT not found: {IMG_ROOT}"

for sp_dir in sorted([p for p in root.iterdir() if p.is_dir()]):
    sp = norm_str(sp_dir.name)
    if sp not in species_to_genus:
        continue
    for img in sp_dir.iterdir():
        if img.is_file() and img.suffix.lower() in IMG_EXT:
            samples.append((str(img), sp))

assert len(samples) > 0, "Không tìm thấy ảnh hợp lệ (hoặc folder species không khớp taxonomy)."
print("Total images:", len(samples))
by_species = {}
for p, sp in samples:
    by_species.setdefault(sp, []).append(p)
```

```python
# cell 3: train/val/test split for each specie
TRAIN_RATIO = 0.70
VAL_RATIO   = 0.15
TEST_RATIO  = 0.15

train_items = []
val_items   = []
test_items  = []

for sp, paths in by_species.items():
    random.shuffle(paths)
    n = len(paths)
    n_train = int(n * TRAIN_RATIO)
    n_val   = int(n * VAL_RATIO)
    n_test = n - n_train - n_val
    train_paths = paths[:n_train]
    val_paths   = paths[n_train:n_train+n_val]
    test_paths  = paths[n_train+n_val:]
    train_items.extend([(p, sp) for p in train_paths])
    val_items.extend([(p, sp) for p in val_paths])
    test_items.extend([(p, sp) for p in test_paths])
print(f"Train images : {len(train_items)}")
print(f"Val images   : {len(val_items)}")
print(f"Test images  : {len(test_items)}")
```

```python
#cell 4: setup label map and enummerate for each iterable
species_list = sorted(list({sp for _, sp in train_items}))
genus_list   = sorted(list({species_to_genus[sp] for sp in species_list}))
family_list  = sorted(list({genus_to_family[g] for g in genus_list}))
order_list   = sorted(list({family_to_order[f] for f in family_list}))

species2idx = {s:i for i,s in enumerate(species_list)}
genus2idx   = {g:i for i,g in enumerate(genus_list)}
family2idx  = {f:i for i,f in enumerate(family_list)}
order2idx   = {o:i for i,o in enumerate(order_list)}

print("order:", len(order2idx), "family:", len(family2idx), "genus:", len(genus2idx), "species:", len(species2idx))
```

```python
# cell 5: Dataset preprocessing & labling id sp, genus, faily , order for each iamge in dataset
class Hier4Dataset(Dataset):
    def __init__(self, items, tf):
        self.items = items
        self.tf = tf

    def __len__(self):
        return len(self.items)

    def __getitem__(self, i):
        path, sp = self.items[i]
        img = Image.open(path).convert("RGB")
        x = self.tf(img)

        g = species_to_genus[sp]
        f = genus_to_family[g]
        o = family_to_order[f]

        y_s = torch.tensor(species2idx[sp], dtype=torch.long)
        y_g = torch.tensor(genus2idx[g], dtype=torch.long)
        y_f = torch.tensor(family2idx[f], dtype=torch.long)
        y_o = torch.tensor(order2idx[o], dtype=torch.long)
        return x, y_o, y_f, y_g, y_s

train_tf = transforms.Compose([
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(15),
    transforms.ColorJitter(brightness=0.15, contrast=0.15),
    transforms.ToTensor(),
    transforms.Normalize([0.485,0.456,0.406],[0.229,0.224,0.225]),
])

val_tf = transforms.Compose([
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize([0.485,0.456,0.406],[0.229,0.224,0.225]),
])

train_loader = DataLoader(
    Hier4Dataset(train_items, train_tf),
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=2,
    pin_memory=True
)

val_loader = DataLoader(
    Hier4Dataset(val_items, val_tf),
    batch_size=64,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)
test_loader = DataLoader(
    Hier4Dataset(test_items, val_tf),
    batch_size=64,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)
```

```python
# 6. Model (ViT-B/16 backbone + 4 heads)
class Hier4Net(nn.Module):
    def __init__(self, n_order, n_family, n_genus, n_species, pretrained=True):
        super().__init__()

        self.backbone = timm.create_model(
            "vit_base_patch16_224",
            pretrained=pretrained,
            num_classes=0
        )

        d = self.backbone.num_features   # 768

        self.drop = nn.Dropout(0.3)

        self.head_o = nn.Linear(d, n_order)
        self.head_f = nn.Linear(d, n_family)
        self.head_g = nn.Linear(d, n_genus)
        self.head_s = nn.Linear(d, n_species)

    def forward(self, x):

        feat = self.backbone(x)
        feat = self.drop(feat)

        return (
            self.head_o(feat),
            self.head_f(feat),
            self.head_g(feat),
            self.head_s(feat)
        )
# Build model
model = Hier4Net(
    len(order2idx),
    len(family2idx),
    len(genus2idx),
    len(species2idx),
    pretrained=True
).to(DEVICE)
# Optimizer
LR = 1e-5
WD = 1e-4
ce = nn.CrossEntropyLoss()
opt = torch.optim.AdamW(
    model.parameters(),
    lr=LR,
    weight_decay=WD
)

scaler = torch.cuda.amp.GradScaler(
    enabled=(DEVICE == "cuda")
)
# Validation
@torch.no_grad()
def evaluate():

    model.eval()

    tot = 0
    co = cf = cg = cs = 0

    for x, y_o, y_f, y_g, y_s in val_loader:

        x = x.to(DEVICE)

        y_o = y_o.to(DEVICE)
        y_f = y_f.to(DEVICE)
        y_g = y_g.to(DEVICE)
        y_s = y_s.to(DEVICE)

        lo, lf, lg, ls = model(x)

        tot += x.size(0)

        co += (lo.argmax(1) == y_o).sum().item()
        cf += (lf.argmax(1) == y_f).sum().item()
        cg += (lg.argmax(1) == y_g).sum().item()
        cs += (ls.argmax(1) == y_s).sum().item()

    return (
        co / tot,
        cf / tot,
        cg / tot,
        cs / tot
    )
# Check model
total_params = sum(p.numel() for p in model.parameters())
trainable_params = sum(
    p.numel()
    for p in model.parameters()
    if p.requires_grad
)

print("Feature dim:", model.backbone.num_features)
print(f"Total params: {total_params:,}")
print(f"Trainable params: {trainable_params:,}")
```

```python
# 7 Train
best_species = -1.0
for epoch in range(1, EPOCHS+1):
    model.train()
    running = 0.0

    for x, y_o, y_f, y_g, y_s in train_loader:
        x   = x.to(DEVICE, non_blocking=True)
        y_o = y_o.to(DEVICE, non_blocking=True)
        y_f = y_f.to(DEVICE, non_blocking=True)
        y_g = y_g.to(DEVICE, non_blocking=True)
        y_s = y_s.to(DEVICE, non_blocking=True)

        opt.zero_grad(set_to_none=True)

        with torch.cuda.amp.autocast(enabled=(DEVICE=="cuda")):
            lo, lf, lg, ls = model(x)
            loss = (
                W_O * ce(lo, y_o) +
                W_F * ce(lf, y_f) +
                W_G * ce(lg, y_g) +
                W_S * ce(ls, y_s)
            )

        scaler.scale(loss).backward()
        scaler.step(opt)
        scaler.update()
        running += loss.item() * x.size(0)

    train_loss = running / max(1, len(train_items))
    acc_o, acc_f, acc_g, acc_s = evaluate()

    print(
        f"Epoch {epoch:02d}/{EPOCHS} | loss={train_loss:.4f} | "
        f"VAL: order={acc_o:.3f} family={acc_f:.3f} genus={acc_g:.3f} species={acc_s:.3f}"
    )

    if acc_s > best_species:
        best_species = acc_s
        torch.save({
            "model": model.state_dict(),
            "order2idx": order2idx,
            "family2idx": family2idx,
            "genus2idx": genus2idx,
            "species2idx": species2idx,
            "species_to_genus": species_to_genus,
            "genus_to_family": genus_to_family,
            "family_to_order": family_to_order,
            "note": "BASE VNUA trained with CORRECT taxonomy mapping (Order/Family/Genus/Species)."
        }, SAVE_PATH)
        print("  -> saved best:", SAVE_PATH)

print("\nDONE. Best VAL species acc:", best_species)
print("Checkpoint:", SAVE_PATH)
```

```python
# cell 8: draw confusion matrix from trained model(species)
model_saved = "/kaggle/working/trained_model_vnua_base_ViT.pt"
best_ckpt = torch.load(model_saved, map_location=DEVICE)

best_model = Hier4Net(
    len(best_ckpt["order2idx"]),
    len(best_ckpt["family2idx"]),
    len(best_ckpt["genus2idx"]),
    len(best_ckpt["species2idx"]),
    pretrained=False
).to(DEVICE)

best_model.load_state_dict(best_ckpt["model"])
best_model.eval()

species2idx_ckpt = best_ckpt["species2idx"]
idx2species = {v: k for k, v in species2idx_ckpt.items()}
species_names = [idx2species[i] for i in range(len(idx2species))]

y_true = []
y_pred = []

with torch.no_grad():
    for x, y_o, y_f, y_g, y_s in test_loader:
        x = x.to(DEVICE, non_blocking=True)
        _, _, _, ls = best_model(x)
        pred_s = ls.argmax(dim=1).cpu().numpy()

        y_true.extend(y_s.numpy())
        y_pred.extend(pred_s)

y_true = np.array(y_true)
y_pred = np.array(y_pred)

species_acc = (y_true == y_pred).mean()
print(f"\nTEST Species accuracy: {species_acc:.4f}")

cm = confusion_matrix(y_true, y_pred, normalize="true")

fig, ax = plt.subplots(figsize=(18, 18))
disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=species_names
)
disp.plot(
    ax=ax,
    cmap="Blues",
    xticks_rotation=90,
    values_format=".2f",
    colorbar=False
)

plt.title("Species-level Confusion Matrixon test dataset")
plt.tight_layout()
plt.savefig(CM_SAVE_PATH, dpi=300, bbox_inches="tight")
plt.show()
```

```python
#TEST EVALUATION (ORDER / FAMILY / GENUS / SPECIES)
import torch
import numpy as np
import pandas as pd

from sklearn.metrics import (
    accuracy_score,
    precision_recall_fscore_support
)
MODEL_PATH = "/kaggle/working/trained_model_vnua_base_ViT.pt"
# LOAD MODEL
ckpt = torch.load(
    MODEL_PATH,
    map_location=DEVICE
)

best_model = Hier4Net(
    len(ckpt["order2idx"]),
    len(ckpt["family2idx"]),
    len(ckpt["genus2idx"]),
    len(ckpt["species2idx"])
).to(DEVICE)

best_model.load_state_dict(ckpt["model"])
best_model.eval()
# STORE LABELS
y_true_o, y_pred_o = [], []
y_true_f, y_pred_f = [], []
y_true_g, y_pred_g = [], []
y_true_s, y_pred_s = [], []
# INFERENCE

with torch.no_grad():

    for x, y_o, y_f, y_g, y_s in test_loader:

        x = x.to(DEVICE)

        lo, lf, lg, ls = best_model(x)

        po = lo.argmax(dim=1).cpu().numpy()
        pf = lf.argmax(dim=1).cpu().numpy()
        pg = lg.argmax(dim=1).cpu().numpy()
        ps = ls.argmax(dim=1).cpu().numpy()

        y_true_o.extend(y_o.numpy())
        y_pred_o.extend(po)

        y_true_f.extend(y_f.numpy())
        y_pred_f.extend(pf)

        y_true_g.extend(y_g.numpy())
        y_pred_g.extend(pg)

        y_true_s.extend(y_s.numpy())
        y_pred_s.extend(ps)

# METRIC FUNCTION
def compute_metrics(y_true, y_pred):

    acc = accuracy_score(
        y_true,
        y_pred
    )

    p_macro, r_macro, f1_macro, _ = \
        precision_recall_fscore_support(
            y_true,
            y_pred,
            average="macro",
            zero_division=0
        )

    _, _, f1_weighted, _ = \
        precision_recall_fscore_support(
            y_true,
            y_pred,
            average="weighted",
            zero_division=0
        )

    return {
        "Accuracy": acc,
        "Precision(Macro)": p_macro,
        "Recall(Macro)": r_macro,
        "F1(Macro)": f1_macro,
        "F1(Weighted)": f1_weighted
    }
# COMPUTE RESULTS

results = pd.DataFrame({
    "Order":   compute_metrics(y_true_o, y_pred_o),
    "Family":  compute_metrics(y_true_f, y_pred_f),
    "Genus":   compute_metrics(y_true_g, y_pred_g),
    "Species": compute_metrics(y_true_s, y_pred_s)
}).T

# DISPLAY

print("\n" + "="*70)
print("TEST PERFORMANCE")
print("="*70)

display(results.round(4))


# SAVE CSV

csv_path = "/kaggle/working/test_metrics_all_taxonomy.csv"

results.to_csv(csv_path)

print("\nSaved:", csv_path)
```
